```python
# Locate the FinMark project folders
project_folder = Path.cwd()

# This handles whether the notebook starts from the project or notebooks folder
if project_folder.name == "notebooks":
    project_folder = project_folder.parent

raw_data_folder = project_folder / "data_raw"
cleaned_data_folder = project_folder / "data_cleaned"

print("Project folder:", project_folder)
print("Raw data folder exists:", raw_data_folder.exists())
print("Cleaned data folder exists:", cleaned_data_folder.exists())

print("\nFiles found:")
for file in raw_data_folder.glob("*.xlsx"):
    print("-", file.name)
```

```python
    # Create working copies to protect the original datasets
demographics_clean = demographics_df.copy()
transactions_clean = transactions_df.copy()
social_media_clean = social_media_df.copy()

print("Rows before removing exact duplicates:")
print("Demographics:", len(demographics_clean))
print("Transactions:", len(transactions_clean))
print("Social media:", len(social_media_clean))

# Remove rows where every column is identical
demographics_clean = demographics_clean.drop_duplicates().copy()
transactions_clean = transactions_clean.drop_duplicates().copy()
social_media_clean = social_media_clean.drop_duplicates().copy()

print("\nRows after removing exact duplicates:")
print("Demographics:", len(demographics_clean))
print("Transactions:", len(transactions_clean))
print("Social media:", len(social_media_clean))

print("\nRows removed:")
print("Demographics:", len(demographics_df) - len(demographics_clean))
print("Transactions:", len(transactions_df) - len(transactions_clean))
print("Social media:", len(social_media_df) - len(social_media_clean))
```

```python
# Check whether the main identifier of each dataset is still duplicated
id_checks = {
    "Customer Demographics": (demographics_clean, "CustomerID"),
    "Customer Transactions": (transactions_clean, "TransactionID"),
    "Social Media Interactions": (social_media_clean, "InteractionID")
}

id_summary = []

for dataset_name, (df, id_column) in id_checks.items():
    duplicated_mask = df.duplicated(subset=id_column, keep=False)

    id_summary.append({
        "Dataset": dataset_name,
        "ID Column": id_column,
        "Rows with Duplicated ID": duplicated_mask.sum(),
        "Number of Duplicated IDs": df.loc[
            duplicated_mask, id_column
        ].nunique()
    })

display(pd.DataFrame(id_summary))
```

```python
# Keep the first occurrence of every primary identifier
demographics_clean = demographics_clean.drop_duplicates(
    subset="CustomerID",
    keep="first"
).reset_index(drop=True)

transactions_clean = transactions_clean.drop_duplicates(
    subset="TransactionID",
    keep="first"
).reset_index(drop=True)

social_media_clean = social_media_clean.drop_duplicates(
    subset="InteractionID",
    keep="first"
).reset_index(drop=True)

print("Rows after resolving duplicated primary IDs:")
print("Demographics:", len(demographics_clean))
print("Transactions:", len(transactions_clean))
print("Social media:", len(social_media_clean))

print("\nRemaining duplicated primary IDs:")
print(
    "CustomerID:",
    demographics_clean["CustomerID"].duplicated().sum()
)
print(
    "TransactionID:",
    transactions_clean["TransactionID"].duplicated().sum()
)
print(
    "InteractionID:",
    social_media_clean["InteractionID"].duplicated().sum()
)
```

```python
# Store the cleaned working datasets for another quality check
clean_datasets = {
    "Customer Demographics": demographics_clean,
    "Customer Transactions": transactions_clean,
    "Social Media Interactions": social_media_clean
}

# Display the remaining missing values after duplicate removal
for dataset_name, df in clean_datasets.items():
    print(f"\n{dataset_name}")
    print(df.isna().sum())
```

```python
# Convert Age and Amount temporarily for inspection
age_check = pd.to_numeric(
    demographics_clean["Age"],
    errors="coerce"
)

amount_check = pd.to_numeric(
    transactions_clean["Amount"],
    errors="coerce"
)

# Identify invalid text values
invalid_age_text = demographics_clean.loc[
    demographics_clean["Age"].notna() & age_check.isna(),
    "Age"
].value_counts()

invalid_amount_text = transactions_clean.loc[
    transactions_clean["Amount"].notna() & amount_check.isna(),
    "Amount"
].value_counts()

print("Invalid Age text:")
print(invalid_age_text)

print("\nSuspicious numeric ages:")
print(
    age_check[
        (age_check < 18) | (age_check > 100)
    ].value_counts().sort_index()
)

print("\nInvalid Amount text:")
print(invalid_amount_text)

print("\nZero or negative transaction amounts:")
print(
    amount_check[
        amount_check <= 0
    ].value_counts().sort_index()
)

# Inspect the categories, including missing values
category_checks = [
    (demographics_clean, "Gender"),
    (demographics_clean, "IncomeLevel"),
    (transactions_clean, "ProductCategory"),
    (transactions_clean, "PaymentMethod"),
    (social_media_clean, "Platform"),
    (social_media_clean, "InteractionType"),
    (social_media_clean, "Sentiment")
]

for df, column in category_checks:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False))
```

```python
# Convert Age to numeric and turn invalid entries into missing values
demographics_clean["Age"] = pd.to_numeric(
    demographics_clean["Age"],
    errors="coerce"
)

demographics_clean.loc[
    ~demographics_clean["Age"].between(18, 100),
    "Age"
] = pd.NA

# Calculate the median using only valid ages
median_age = demographics_clean["Age"].median()

# Replace missing and invalid ages with the median
demographics_clean["Age"] = (
    demographics_clean["Age"]
    .fillna(median_age)
    .astype("Int64")
)

# Preserve customers with unknown income instead of guessing a category
demographics_clean["IncomeLevel"] = (
    demographics_clean["IncomeLevel"]
    .fillna("Unknown")
    .str.strip()
)

# Standardize other text columns
demographics_clean["CustomerID"] = (
    demographics_clean["CustomerID"].str.strip()
)

demographics_clean["Gender"] = (
    demographics_clean["Gender"].str.strip()
)

demographics_clean["Location"] = (
    demographics_clean["Location"].str.strip()
)

# Convert mixed date formats into a proper date data type
demographics_clean["SignupDate"] = pd.to_datetime(
    demographics_clean["SignupDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Median age used:", median_age)
print("\nMissing values after demographics cleaning:")
print(demographics_clean.isna().sum())

print("\nDemographics data types:")
print(demographics_clean.dtypes)

print("\nAge range:")
print(
    demographics_clean["Age"].min(),
    "to",
    demographics_clean["Age"].max()
)

print("\nIncome level values:")
print(demographics_clean["IncomeLevel"].value_counts())
```

```python
# Convert Amount to numeric and turn text such as "Free" into missing values
transactions_clean["Amount"] = pd.to_numeric(
    transactions_clean["Amount"],
    errors="coerce"
)

# Treat zero and negative transaction amounts as invalid
transactions_clean.loc[
    transactions_clean["Amount"] <= 0,
    "Amount"
] = pd.NA

# Calculate the median using only valid positive amounts
median_amount = transactions_clean["Amount"].median()

# Replace missing and invalid amounts with the median
transactions_clean["Amount"] = (
    transactions_clean["Amount"]
    .fillna(median_amount)
    .round(2)
)

# Preserve transactions with an unknown category
transactions_clean["ProductCategory"] = (
    transactions_clean["ProductCategory"]
    .fillna("Unknown")
    .str.strip()
)

# Standardize text columns
transactions_clean["CustomerID"] = (
    transactions_clean["CustomerID"].str.strip()
)

transactions_clean["TransactionID"] = (
    transactions_clean["TransactionID"].str.strip()
)

transactions_clean["PaymentMethod"] = (
    transactions_clean["PaymentMethod"].str.strip()
)

# Convert mixed date formats into a proper date data type
transactions_clean["TransactionDate"] = pd.to_datetime(
    transactions_clean["TransactionDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Median amount used:", median_amount)

print("\nMissing values after transaction cleaning:")
print(transactions_clean.isna().sum())

print("\nTransaction data types:")
print(transactions_clean.dtypes)

print("\nTransaction amount range:")
print(
    transactions_clean["Amount"].min(),
    "to",
    transactions_clean["Amount"].max()
)

print("\nProduct category values:")
print(transactions_clean["ProductCategory"].value_counts())
```

```python
# Preserve records with missing platforms or sentiments
social_media_clean["Platform"] = (
    social_media_clean["Platform"]
    .fillna("Unknown")
    .str.strip()
)

social_media_clean["Sentiment"] = (
    social_media_clean["Sentiment"]
    .fillna("Unknown")
    .str.strip()
)

# Standardize sentiment categories while preserving their direction
sentiment_mapping = {
    "Very Positive": "Positive",
    "Very Negative": "Negative"
}

social_media_clean["Sentiment"] = (
    social_media_clean["Sentiment"]
    .replace(sentiment_mapping)
)

# Standardize other text columns
social_media_clean["CustomerID"] = (
    social_media_clean["CustomerID"].str.strip()
)

social_media_clean["InteractionID"] = (
    social_media_clean["InteractionID"].str.strip()
)

social_media_clean["InteractionType"] = (
    social_media_clean["InteractionType"].str.strip()
)

# Convert mixed date formats into a proper date data type
social_media_clean["InteractionDate"] = pd.to_datetime(
    social_media_clean["InteractionDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Missing values after social media cleaning:")
print(social_media_clean.isna().sum())

print("\nSocial media data types:")
print(social_media_clean.dtypes)

print("\nPlatform values:")
print(social_media_clean["Platform"].value_counts())

print("\nInteraction type values:")
print(social_media_clean["InteractionType"].value_counts())

print("\nSentiment values:")
print(social_media_clean["Sentiment"].value_counts())
```

```python
# Rebuild the dictionary using the fully cleaned versions
clean_datasets = {
    "Customer Demographics": demographics_clean,
    "Customer Transactions": transactions_clean,
    "Social Media Interactions": social_media_clean
}

# Validate rows, missing values, and exact duplicates
validation_summary = []

for dataset_name, df in clean_datasets.items():
    validation_summary.append({
        "Dataset": dataset_name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing Values": df.isna().sum().sum(),
        "Exact Duplicates": df.duplicated().sum()
    })

display(pd.DataFrame(validation_summary))

# Validate primary identifiers
print("Primary ID validation:")
print(
    "Duplicated CustomerID:",
    demographics_clean["CustomerID"].duplicated().sum()
)
print(
    "Duplicated TransactionID:",
    transactions_clean["TransactionID"].duplicated().sum()
)
print(
    "Duplicated InteractionID:",
    social_media_clean["InteractionID"].duplicated().sum()
)

# Validate numeric values
print("\nNumeric validation:")
print(
    "All ages are between 18 and 100:",
    demographics_clean["Age"].between(18, 100).all()
)
print(
    "All transaction amounts are positive:",
    transactions_clean["Amount"].gt(0).all()
)

# Validate date data types
print("\nDate validation:")
print(
    "SignupDate is a date:",
    pd.api.types.is_datetime64_any_dtype(
        demographics_clean["SignupDate"]
    )
)
print(
    "TransactionDate is a date:",
    pd.api.types.is_datetime64_any_dtype(
        transactions_clean["TransactionDate"]
    )
)
print(
    "InteractionDate is a date:",
    pd.api.types.is_datetime64_any_dtype(
        social_media_clean["InteractionDate"]
    )
)

# Validate connections between datasets
demographic_customer_ids = set(
    demographics_clean["CustomerID"]
)

transaction_unmatched = (
    ~transactions_clean["CustomerID"].isin(
        demographic_customer_ids
    )
).sum()

social_unmatched = (
    ~social_media_clean["CustomerID"].isin(
        demographic_customer_ids
    )
).sum()

print("\nCustomerID connection validation:")
print(
    "Transaction customers not found in demographics:",
    transaction_unmatched
)
print(
    "Social media customers not found in demographics:",
    social_unmatched
)
```

```python
# ==========================================
# EDA EVIDENCE: Relationships, Anomalies & ML Readiness
# ==========================================

from pathlib import Path
import pandas as pd


project_folder = Path.cwd()

# 1. Load all three cleaned datasets
df_demo = pd.read_csv(project_folder / "customer_demographics_cleaned.csv")
df_trans = pd.read_csv(project_folder / "customer_transactions_cleaned.csv")
df_social = pd.read_csv(project_folder / "social_media_interactions_cleaned.csv")

print("==================================================")
print("   RELATIONSHIPS, ANOMALIES & ML READINESS EVIDENCE ")
print("==================================================")

# 2. Dataset Characteristics & ML Readiness (Shape & Missing/Unknown tracking)
print("\n--- 1. DATASET CHARACTERISTICS & ML READINESS ---")
datasets = {
    "Customer Demographics": df_demo,
    "Customer Transactions": df_trans,
    "Social Media Interactions": df_social
}

for name, df in datasets.items():
    print(f"\nDataset: {name}")
    print(f"  - Shape (Rows, Columns): {df.shape}")
    # Count nulls and preserved 'Unknown' string placeholders
    unresolved_counts = (df == 'Unknown').sum() + df.isnull().sum()
    active_unresolved = unresolved_counts[unresolved_counts > 0]
    if not active_unresolved.empty:
        print("  - Unresolved / Placeholder Counts:")
        print(active_unresolved.to_string())
    else:
        print("  - No missing or placeholder values found.")

# 3. Outlier & Anomaly Verification (Numerical boundaries)
print("\n--- 2. OUTLIER & ANOMALY VERIFICATION ---")
if 'Age' in df_demo.columns:
    print(f"Demographics ('Age') -> Min: {df_demo['Age'].min()}, Max: {df_demo['Age'].max()}, Std Dev: {df_demo['Age'].std():.2f}")
if 'Amount' in df_trans.columns:
    print(f"Transactions ('Amount') -> Min: {df_trans['Amount'].min()}, Max: {df_trans['Amount'].max()}, Std Dev: {df_trans['Amount'].std():.2f}")

# 4. Variable Relationships & Trends (Cross-Tabulation)
print("\n--- 3. VARIABLE RELATIONSHIPS & TRENDS ---")
if 'PaymentMethod' in df_trans.columns and 'ProductCategory' in df_trans.columns:
    print("\nCross-Tabulation: Payment Method vs Product Category")
    cross_tab = pd.crosstab(df_trans['PaymentMethod'], df_trans['ProductCategory'])
    print(cross_tab)

print("\n==================================================")
```
